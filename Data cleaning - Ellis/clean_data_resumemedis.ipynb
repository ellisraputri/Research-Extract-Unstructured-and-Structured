```python
import pandas as pd
import numpy as np
import re

df = pd.read_csv('../data/rsud_gunungjati.resumeMedis.csv')
df.info()
```

```python
# remove every diagnosaDokter[xx].no
pattern = r"^diagnosaDokter\[\d+\]\.no$"

df = df[[col for col in df.columns if not re.match(pattern, col)]]

df.info()
```

```python
# remove every diagnosaDokter[xx] where xx >=3
def is_high_index_diagnosa(col):
    match = re.match(r"^diagnosaDokter\[(\d+)\]\.", col)
    return match and int(match.group(1)) >= 3

df = df[[col for col in df.columns if not is_high_index_diagnosa(col)]]

df.info()
```

```python
# remove every diagnosaDokter[xx].jenisDiagnosa.value 
pattern = r"^diagnosaDokter\[\d+\]\.jenisDiagnosa.value$"

df = df[[col for col in df.columns if not re.match(pattern, col)]]

df.info()
```

```python
# remove every diagnosaDokter[xx].isLoadBtnDiagnosaDokter 
pattern = r"^diagnosaDokter\[\d+\]\.isLoadBtnDiagnosaDokter$"

df = df[[col for col in df.columns if not re.match(pattern, col)]]

df.info()
```

```python
# remove every diagnosaDokter[xx].norecDiagnosa 
pattern = r"^diagnosaDokter\[\d+\]\.norecDiagnosa$"

df = df[[col for col in df.columns if not re.match(pattern, col)]]

df.info()
```

```python
# remove every detailObatResep[xx] where xx >=5
def is_high_index_diagnosa(col):
    match = re.match(r"^detailObatResep\[(\d+)\]\.waktupakai", col)
    return match

df = df[[col for col in df.columns if not is_high_index_diagnosa(col)]]

df.info()
```

```python

def is_high_index_diagnosa(col):
    match = re.match(r"^diagnosaDokter\[(\d+)\]\.diagnosaIcd10.value", col)
    return match 

df = df[[col for col in df.columns if not is_high_index_diagnosa(col)]]

df.info()
```

```python
# remove every detailObatResep[xx] where xx >=5
def is_high_index_diagnosa(col):
    match = re.match(r"^detailObatResep\[(\d+)\]\.obat.label", col)
    return match and int(match.group(1)) >= 3

def is_high_index_tindakan(col):
    match = re.match(r"^detailTindakan\[(\d+)\]\.ketTindakanDokter", col)
    return match and int(match.group(1)) >= 3

def is_high_index_icd9(col):
    match = re.match(r"^detailTindakan\[(\d+)\]\.diagnosaIcd9.label", col)
    return match and int(match.group(1)) >= 3

df = df[[col for col in df.columns if not is_high_index_diagnosa(col)]]
df = df[[col for col in df.columns if not is_high_index_tindakan(col)]]
df = df[[col for col in df.columns if not is_high_index_icd9(col)]]

df.info()
```

```python

def is_match(col):
    matches = [False] * 100
    matches[0] = re.match(r"^detailTindakan\[(\d+)\]\.no", col)
    matches[1] = re.match(r"^detailTindakan\[(\d+)\]\.norecDiagnosa9", col)
    matches[2] = re.match(r"^detailTindakan\[(\d+)\]\.diagnosaIcd9.value", col)
    matches[3] = re.match(r"^detailObatResep\[(\d+)\]\.no", col)
    matches[4] = re.match(r"^detailTindakan\[(\d+)\]\.diagnosaIcd9$", col)
    matches[5] = re.match(r"^detailObatResep\[(\d+)\]\.obat.value", col)
    matches[6] = re.match(r"^detailObatResep\[(\d+)\]\.aturanPakai", col)
    matches[7] = re.match(r"^detailObatResep\[(\d+)\]\.jumlah", col)
    matches[8] = re.match(r"^detailObatResep\[(\d+)\]\.dosis", col)
    matches[9] = re.match(r"^diagnosaDokter\[(\d+)\]\.diagnosaIcd10$", col)
    matches[10] = re.match(r"^detailObatResep\[(\d+)\]\.obat", col)
    matches[11] = re.match(r"^detailTindakan\[(\d+)\]\.norecDiagnosa", col)
    matches[12] = re.match(r"^detailObatResep\[(\d+)\]\.jenisObat", col)
    matches[13] = re.match(r"^detailObatResep\[(\d+)\]\.waktuPakai", col)
    
    match = False
    for m in matches:
        match = (match or m)
    return match

df = df[[col for col in df.columns if not is_match(col)]]

df.info()
```
