```python
import pandas as pd

jantung_df = pd.read_csv('cppt_jantung_extract.csv')
remed_df  = pd.read_csv('resume_medis_extract.csv')

jantung_df = jantung_df.rename(columns={'diagnosaDokter[0].keterangan': 'diagnosa'})
remed_df = remed_df.rename(columns={'diagnosaDokter[0].ketDiagnosaDok':'diagnosa'})

df = (pd.concat([jantung_df, remed_df], ignore_index=True, sort=False).drop_duplicates(subset=['_id']))

df.head()
```

```python
df = df.drop_duplicates(subset=['grouped_diagnosa', 'icd10_label'], keep='first')

df.info()
```

```python
df_sorted = df.sort_values(by='grouped_diagnosa')
df_sorted.head()
```

```python
df_sorted.to_csv('diagnosa_unique_list.csv')
```
