```python
import pandas as pd
import re

df = pd.read_csv("3000_data_gabungan.csv")
df.info()
```

```python
df['keluhan_utama'] = df['S'].combine_first(df['riwayatPenyakit'])
df['pengobatan'] = df['P'].combine_first(df['hasilIntruksi'])

df['beratBadan'] = df['beratBadan'].astype(str).str.replace(',', '.', regex=False)
df['beratBadan'] = pd.to_numeric(df['beratBadan'], errors='coerce')

df['tinggiBadan'] = df['tinggiBadan'].astype(str).str.replace(',', '.', regex=False)
df['tinggiBadan'] = pd.to_numeric(df['tinggiBadan'], errors='coerce')
df['nadi'] = pd.to_numeric(df['nadi'], errors='coerce')

df['suhu'] = df['suhu'].astype(str).str.replace(',', '.', regex=False)
df['suhu'] = pd.to_numeric(df['suhu'], errors='coerce')

df['pernapasan'] = df['pernapasan'].astype(str).str.replace(',', '.', regex=False)
df['pernapasan'] = pd.to_numeric(df['pernapasan'], errors='coerce')
df['SPO2'] = pd.to_numeric(df['SPO2'], errors='coerce')

df['tekananDarah'] = df['tekananDarah'].astype(str).str.replace("\\", "/", regex=False)
df[['tekanan_sistolik', 'tekanan_diastolik']] = df['tekananDarah'].str.split('/', expand=True)
df['tekanan_sistolik'] = pd.to_numeric(df['tekanan_sistolik'], errors='coerce')
df['tekanan_diastolik'] = pd.to_numeric(df['tekanan_diastolik'], errors='coerce')

df.info()
```

```python
pattern = r"^Unnamed:"
df = df[[col for col in df.columns if not re.match(pattern, col)]]

df = df.rename(columns={
    'O' : 'detailPemeriksaan',
    'S' : 'detailKeluhan',
    'P'  : 'detailPengobatan',
})

df.to_csv("final_unextracted_3000data.csv")
```
